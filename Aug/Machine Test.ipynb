```python
bookings={
    102:'vacant',
    103:'vacant',
    104:'vacant',
    105:'vacant'
}
```

```python
while(True):
    choice=int(input(f'Menu:\nHotel Room Booking System\n1. Book Room\n2. View All Bookings\n3. Search Booking\n4. Update Booking Days\n5. Cancel Booking\n6. Exit\nEnter your choice:'))
    if choice==1:
        book_room(bookings)
    elif choice==2:
        view_bookings(bookings)
    elif choice==3:
        search_booking(bookings)
    elif choice==4:
        update_days(bookings)
    elif choice==5:
        cancel_booking(bookings)
    elif choice==6:
        print('Thank You... Program Terminated')
        break
    else:
        print('Invalid Choice !')
    
```

```python
def book_room(bookings):
    rm=int(input('Enter the room number : '))
    gn=input('Enter the Guest name : ')
    rt=input('Enter the room type  : ')
    dy=int(input('Enter the Number of days : '))
    if dy<1:
        dy=int(input('Day should be more than 1, please Enter again'))
    total_price=int(input('Enter the Price : '))
    if total_price<1:
        total_price=int(input('Price Should be more than 0 ,Please Enter the Price Again : '))
    if rm not in bookings:
        print('Bookings Not Found')
    elif bookings[rm]!='vacant':
        print('The Room Number is already booked')
    else:
        bookings[rm]={
    'guest_name':gn,
    'room_type':rt,
    'days':dy,
    'total_price':total_price
    }
    print('Room Booked Successfully.')
    
```

```python
def view_bookings(bookings):
    lent=len(bookings.values())
    if lent==0:
        print('No Bookings Records Found')
    else:
        for i,j in bookings.items():
            print(i,' : ',j)
```

```python
def search_booking(bookings):
    n=int(input('Enter the room number : '))
    if n in bookings:
        print(bookings[n])
    else:
        print('Bookings not Found !')
```

```python
def update_days(bookings):
    rm=int(input('Enter the room number : '))
    if rm not in bookings:
        print('The Room Number is not present')
    else:
        dy=int(input('Enter the Number od days : '))
        bookings[rm].update({'days':dy})
    print('Days Updated Successfully.')
```

```python
def cancel_booking(bookings):
    rm=int(input('Enter the room number : '))
    if rm not in bookings:
        print('The Room Number is not present')
    else:
        del bookings[rm]
    print('Cancelled Booking')
```
