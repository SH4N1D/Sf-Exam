```python
1. Perfect Number – Number Logic
Write a Python program to determine whether a number is a Perfect number by finding the sum of
its proper divisors.
Requirements:
• Read a number from the user.
• Find all proper divisors of the number.
• Calculate the sum of the proper divisors.
• Compare the sum with the original number.
• Display whether the number is a Perfect number or not.
```

```python
n=int(input('Enter the Number : '))
divisors=[]
limit=round((n/2)+1)
for i in range(1,limit):
    if n%i==0:
        divisors.append(i)
print('Divisors :',divisors)
print('Sum : ',sum(divisors))
if sum(divisors)!=n:
    print('Not a Perfect Number')
```

```python
2. Shopping Bill System – Application Logic
A customer purchases multiple products. Store product names, quantities, and prices. Calculate the
total bill and apply a 10% discount if the bill exceeds ■5,000.
Requirements:
• Read the number of products from the user.
• Store the product name, quantity, and price.
• Calculate the total price for each product.
• Calculate the total bill amount.
• Apply a 10% discount if the total exceeds ■5,000.
• Display the final bill amount.
```

```python
product=['a','b','c','d']
count=[20,30,43,32]
prices=[1200,300,250,100]
amount={}
print(f'Product list :{product}\nQuantity : {count}\nPrices : {prices}')
n=int(input('Enter the Number of Products You want to make purchase : '))
for i in range(n):
    name=input('Enter the Porduct Name :')
    qty=int(input('Enter the Quantity : '))
    p=product.index(name)
    if name in amount:
        amount[name]=amount[name]+(qty*prices[p])
    else:
        amount[name]=qty*prices[p]
print('Total Purchase of each product :')
for x,y in amount.items():
    print(x,':',y)
total=sum(amount.values())
print('Total : ',total)
if total>5000:
    total=total-(total*.10)
    print('After Applying Discount : ',total)
```

```python
3. Pattern Printing – X Pattern
Write a Python program to print the following X pattern using nested loops:
*   *
 * *
  *
 * *
*   *
Requirement: The program should accept the size of the pattern from the user and generate the
pattern accordingly.
```

```python
n=int(input('Enter the Value : '))
for i in range(n-1):
    for j in range(i):
        print(' ',end='')
    for k in range(n,i,-1):
        if k==n or k==(i+1):
            print(' * ',end='')
        else:
            print('   ',end='')
    print()
for i in range(n):
    for j in range(n-1,i,-1):
        print(' ',end='')
    for k in range(-1,i):
        if k==-1 or k==(i-1):
            print(' * ',end='')
        else:
            print('   ',end='')
    print()
```
