### SECTION E — Matplotlib Visualization (6 Marks)


Q18. Line Chart [2 marks]
Create a line chart showing total monthly sales for each department. Requirements: X-axis → Department;
Y-axis → Total Monthly Sales; add chart title, X-axis label, Y-axis label, markers, grid, and a legend.
Suggested title: Department-wise Monthly Sales.


```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,5))
plt.plot(final_dataFrame['Department'],final_dataFrame['Monthly_Sales'],marker='o',label='Department vs Sales')
plt.xlabel('Department')
plt.ylabel('Total Monthly Sales')
plt.legend()
plt.grid()
plt.title('Department-wise Monthly Sales')
plt.show()
```

```python
Q19. Bar Chart [2 marks]
Create a bar chart showing the average basic salary for each department. Requirements: X-axis →
Department; Y-axis → Average Basic Salary; add title, X-axis label, Y-axis label, appropriate tick labels, a
legend
```

```python
plt.figure(figsize=(8,5))
plt.bar(final_dataFrame['Department'],final_dataFrame['Basic_Salary'],label='Department vs Sales')
plt.xlabel('Department')
plt.ylabel('Salaries')
plt.legend()
plt.grid()
plt.xticks(rotation=45)
plt.yticks(rotation=35)
plt.title('Department-wise Monthly Sales')
plt.show()
```

Q20. Pie Chart [2 marks]
Create a pie chart showing the percentage distribution of employees across departments. Use
department-wise employee count. Display percentages using autopct, department names, a title, and a
legend. Use an appropriate figure size.

```python
pie=final_dataFrame.groupby('Department')['Employee_ID'].count()
pie
```

```python
plt.figure(figsize=(8,5))
plt.pie(pie.values,labels=pie.index,autopct='%0.1f%%')
plt.title('Distribution Of Employees Across Departments')
plt.show()
```
