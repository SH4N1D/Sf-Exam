Question 1 — Number Pattern & Logic (10 Marks)


Write a Python program to print the following pattern for n = 5:
1
12
123
1234
12345
Requirements: Take n as input from the user, use nested for loops, and make the program work for
any positive value of n.

```python
n=int(input('Enter the Value : '))
if n>0:
    for i in range(1,n+1):
        for j in range(1,i+1):
            print(j,end='')
        print()
else:
    print('the program works only for any positive value of n')
```

### Question 2 — Logical Number Processing (10 Marks)


Write a Python program to find the second largest unique number in a list without using the built-in
sort() or sorted() functions.
Given: numbers = [10, 25, 8, 25, 40, 15, 40, 30]
Expected output: Second Largest Number: 30
Requirements: Handle duplicate values and display an appropriate message if there is no
second-largest unique number.

```python
l=list(map(int,input('Enter the list of Numbers : ').split()))
s=-9999
a=max(l)
l=list(set(l))
for i in l:
    if (i>s) and i!=a:
        s=i
print('Second Largest : ',s)
```

## Question 3 — String Logic (10 Marks)


Write a Python program to find the first non-repeating character in a string.
Given: text = "swiss"
Expected output: First Non-Repeating Character: w
For text = "aabbcc", display: No non-repeating character found.
Requirements: Do not use external libraries, use loops and appropriate Python data structures, and
make the program case-insensitive.

```python
a=input('Enter the String : ')
di={}
for i in a:
    if i in di:
        di[i]+=1
    else:
        di[i]=1
for x,y in di.items():
    if y==1:
        print(x,end='')
```
